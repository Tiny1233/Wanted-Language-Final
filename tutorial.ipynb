# Welcome to **Wanted Language Tutorial**
## 1. Getting Started
### 1.1 Starting

Wanted is a language made by Tiny1233 from 2026/7/6, it can be used to run a sort of codes. Now, you can create your first Wanted File and name it `*.wcn`. Built-in package is not defaultly included, so you can use keyword `ext` to load it. Then write the output. The following is the whole code:

```python
ext @builtin.wcc
output('Hello, World!')
```

Or use `getwcore` sentence. But remember, the `getwcore` is much longer and it'll lead to unexpected errors if you don't understand the sentences. The output function is defined in this way:

```python
func @override output(cont: str, sep = '\n'):
    return (
        getwcore 'stdout' cont + sep
    )
```

like one in `C` (stdio.h):

```python
int printf (const char *__format, ...)
{
  register int __retval;
  __builtin_va_list __local_argv; __builtin_va_start( __local_argv, __format );
  __retval = __mingw_vprintf( __format, __local_argv );
  __builtin_va_end( __local_argv );
  return __retval;
}
```
